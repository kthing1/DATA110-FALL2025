### 3. Multiple Lines

- Create a plot with the following three functions on the same graph:
 - $f(x) = x^3$
 - $g(x) = e^x$
 - $h(x) = \log ⁡(x)$ (for $x > 0$)
- Use different colors for each line
- Add a legend to distinguish between the functions


````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2,4,500)

f_3=pow(x,3)
plt.plot(x,f_3,color='#F25278',label='$f(x)=x^3$')

g_3=np.e**x
plt.plot(x,g_3,color='#FC4C4E',label='$g(x)=e^x$')

x_clipped=np.clip(x, 0.001, None)
h_3=np.log(x_clipped)
plt.plot(x_clipped,h_3,color='#F79AC0',label='$h(x)=log(x)$')

plt.legend()

plt.show()
````

### 8. Customizing Scatter Plots

- Generate data for the function $f(x) = x sin(x)$ over the range [ 0 , 4π ]
- Create a scatter plot of this data with a line connecting the points
- Customize the plot by changing the marker size, line width, marker face color, marker edge color, and marker edge width


````python
x_8=np.linspace(0,4*np.pi)

f_8=x_8*np.sin(x_8)
plt.plot(x_8,f_8,'-o', color='k',
         markersize=10, linewidth=2,
         markerfacecolor='#ff1694',
         markeredgecolor='k',
         markeredgewidth=2)

plt.title(" $f(x) = x sin(x)$ ")

plt.show()
````

#### Notes on my selected functions:
I decided to plot $tan(x)$ and $arctan(x)$ together because they are inverse functions, and I was curious to see how their graphs contrast. I did not limit the number of points plotted because when I tested it earlier smaller sets such as 15 and 37 pointsm the vertical asymptotes did not appear clearly.

As for my styling, I decided to go with the `dark_background` plot style because the dark plot would stand out from the light `seaborn` style used everywhere else in this assignment.

````python
x_10=np.linspace(-1*np.pi,1*np.pi)

f_10=np.tan(x_10)
plt.plot(x_10,f_10,':oc',
         markersize=5, linewidth=2,
         markerfacecolor='c',
         markeredgecolor='b',
         markeredgewidth=1,
         label='$tan(x)$')

i_10=np.arctan(x_10)
plt.plot(x_10,i_10,'-^y',
         markersize=5, linewidth=1,
         markerfacecolor='y',
         markeredgecolor='y',
         markeredgewidth=1,
         label='$arctan(x)$')

plt.ylim(-10,10)

plt.xlabel('x')
plt.ylabel('y')
plt.title(" Comparison of TAN and ARCTAN ", fontsize=18)

plt.legend();
````
