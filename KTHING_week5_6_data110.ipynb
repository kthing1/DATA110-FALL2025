# Task 1:
The dataset shows information about Titanic passengers, including whether they survived, their age, sex, and if they were traveling alone. It also includes where they departed from and how many family members they had on board such as sibsp (siblings/spouses) and parch (parents/children). We can see each passenger’s class and fare, with 1st class being the most expensive and 3rd class the cheapest. The dataset also notes whether a passenger was an adult male. These columns are important because they capture key aspects such as socioeconomic status, gender, age, and family connections, all of which may have affected survival chances. By examining these variables, we can explore how wealth, gender, and age were related to outcomes during the disaster.

```python
# Load dataset
titanic = sns.load_dataset('titanic')

# Histogram + KDE
plt.figure(figsize=(8,5))
sns.histplot(data=titanic, x='fare', kde=True, bins=40, color='teal')
plt.title('Fare Distribution of Titanic Passengers')
plt.xlabel('Fare (British Pounds)')
plt.ylabel('Count')
plt.show()
```

# Task 2:
The fare distribution is clearly right-skewed, showing that most passengers paid low fares while a few paid much higher amounts. The long right tail represents outliers, likely wealthier first-class passengers. This pattern shows strong socioeconomic differences on board, with many low-income travelers and a smaller group of rich ones.

```python
plt.figure(figsize=(8,5))
sns.kdeplot(data=titanic, x='fare', hue='survived', fill=True, common_norm=False)
plt.title('Fare Distributions by Survival Status')
plt.xlabel('Fare (British Pounds)')
plt.ylabel('Density')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

# Task 3:
Passengers who survived generally paid higher fares, with their KDE curve appearing more to the right. Those who did not survive mostly paid lower fares, typical of third-class passengers. The difference in the two distributions suggests that wealth and class affected survival, as higher-paying passengers had better access to lifeboats and safer areas of the ship.

```python
plt.figure(figsize=(8,6))
sns.boxplot(data=titanic, x='pclass', y='fare', hue='sex')
plt.title('Fare by Passenger Class and Sex')
plt.xlabel('Passenger Class')
plt.ylabel('Fare (£)')
plt.show()
```

# Task 4:
Median fare goes down from 1st to 3rd class, showing differences in wealth. Males and females in the same class paid similar fares. 1st class has bigger spreads and more outliers, showing richer passengers paid much more.

```python
plt.figure(figsize=(8,6))
sns.violinplot(data=titanic, x='sex', y='age', inner='box')
plt.title('Age Distribution by Sex')
plt.xlabel('Sex')
plt.ylabel('Age')
plt.show()
```

# Task 5:
Age distributions are similar for males and females, but males vary a bit more. Most passengers are adults between 20–40, with some children and older people. The plot shows the main ages and how spread out they are.

```python
plt.figure(figsize=(8,6))
sns.violinplot(data=titanic, x='sex', y='age', inner=None, color='lightgray')
sns.stripplot(data=titanic, x='sex', y='age', jitter=True, alpha=0.4)
plt.title('Age by Sex with Individual Observations')
plt.xlabel('Sex')
plt.ylabel('Age')
plt.show()
```

# Task 6:
Adding points shows clusters and outliers that the violin hides. You can see exactly where children and older passengers are. It reveals details about individual ages not clear from the overall shape
